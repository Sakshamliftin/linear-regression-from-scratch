```python
#import libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
#create sythetic data

true_w = 2
true_b = 3

x = np.linspace(1,10,100)
y = true_w*x+true_b + np.random.normal(0,10,100)

# define the model

w = 1
b=2

def model(x,w,b):
  return w*x+b


def mse(y,y_hat):
    return np.mean((y-y_hat)**2)

#update w and b values

def updateParams(w,b,x,y,y_hat):

  temp_w = w - 0.05*(np.mean((y_hat-y)*x))
  temp_b = b - 0.05*(np.mean(y_hat-y))
  w = temp_w
  b = temp_b
  return w,b


# loop for gradient descent

losses = []
for i in range(100):
    y_hat = model(x, w, b)
    loss = mse(y, y_hat)
    print(loss)
    losses.append(loss)
    w, b = updateParams(w, b, x, y, y_hat)

# Plot loss vs iterations
plt.plot(losses)
plt.title("Loss over Iterations")
plt.xlabel("Iterations")
plt.ylabel("Loss")
plt.show()

plt.scatter(x, y, color='blue', label='Data')
plt.plot(x, model(x, w, b), color='red', label='Fitted Line')
plt.legend()
plt.show()


print("w value is:",w)
print("b value is",b)
```
